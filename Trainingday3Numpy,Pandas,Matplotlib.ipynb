# Numpy

```python
import numpy as np
```

```python
students =np.array([
    [85,90,78],
    [60,55,70],
    [95,92,88],
    [40,50,45],
    [72,68,75]
])
students
```

```python
avg_marks = np.mean(students, axis=0)
print("Average marks of each student:", avg_marks)
```

```python
avg_marks = np.mean(students, axis=1)
print("Average marks of each student:", avg_marks)
```

```python
topper = np.argmax(avg_marks)
print("topper students:",topper)
```

```python
failed_student = np.where(avg_marks<50)
print("failed students:",failed_student[0]+1)
```

```python
for i, avg in enumerate(avg_marks):
  status = "pass" if avg > 50 else "fail"
  print(f"student(1+1): Avg = {avg:.2f} | {status} ")
```

```python
for i, avg in enumerate(avg_marks):
  status = "pass" if avg > 50 else "fail"
  print(f"student(1+1): Avg = {avg:.2f} / {status} ")
```

# ***Matplotlib***

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [2,4,6,8,10]
plt.plot(x,y)
```

```python
x = [1,2,3,4,5]
y = [2,4,6,8,10]
plt.plot(x,y)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Line Plot')
plt.grid(True)
plt.show()
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
temperature = [25, 35, 40 ,50,60,70,80,90,100,110,120,130]
plt.bar(months, temperature, color='pink')
plt.xlabel('Months')
plt.ylabel('Temperature')
plt.title('temperature Plot')

plt.show()
```

```python
plt.barh(months, temperature, color='pink')
plt.xlabel('Months')
plt.ylabel('Temperature')
plt.title('temperature Plot (Horizontal)')

plt.show()
```

```python
height = [150, 160, 170, 180, 190]
weight = [50, 60, 70, 80, 90]
plt.scatter(height, weight, color='pink')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.title('height vs weight')

plt.show()
```

```python
brands = ['Brand A', 'Brand B', 'Brand C']
sales = [100, 150, 200]
plt.pie(sales, labels=brands , autopct = '%1.1f%%', startangle=90, shadow=True)
plt.title('Sales by Brand')
plt.axis('equal')
plt.show()
```

```python
x = [1,2,3,4]
y1 = [2,4,6,8]
y2 = [3,6,9,12]
plt.plot(x,y1, label='square')
plt.plot(x,y2, label='linear')
plt.legend()
plt.title('multiple lines')
plt.show()
```

```python
# Subplots
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# Correcting x and y by using matched lists (x and y1 both have length 4)
axs[0, 0].plot(x, y1, color='blue')
axs[0, 0].set_title('Square/Linear Plot')

# Correcting bar plot dimensions (using brands and sales which both have length 3)
axs[0, 1].bar(brands, sales, color='pink')
axs[0, 1].set_title('Sales')

# Scatter plot (height and weight both have length 5)
axs[1, 0].scatter(height, weight, color='pink')
axs[1, 0].set_title('Scatter')

# Correcting the pie chart to use 'sales' instead of the undefined 'shares' variable
axs[1, 1].pie(sales, labels=brands, autopct='%1.1f%%')
axs[1, 1].set_title('Pie')

plt.tight_layout()
plt.show()
```
