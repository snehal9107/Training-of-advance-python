```python
class Student:
    def __init__(self, r, n, m1, m2, m3, m4, m5):
        self.rollno = r
        self.name = n
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3
        self.m4 = m4
        self.m5 = m5

    def calculate(self):
        total = self.m1 + self.m2 + self.m3 + self.m4 + self.m5
        percentage = total / 5

        print("Roll No:", self.rollno)
        print("Name:", self.name)
        print("Total Marks:", total)
        print("Percentage:", percentage)

        if percentage >= 85:
            print("Student can take admission")
        else:
            print("Student can't take admission")


r = int(input("Enter roll no: "))
n = input("Enter name: ")

m1 = int(input("Enter marks of paper 1: "))
m2 = int(input("Enter marks of paper 2: "))
m3 = int(input("Enter marks of paper 3: "))
m4 = int(input("Enter marks of paper 4: "))
m5 = int(input("Enter marks of paper 5: "))

s = Student(r, n, m1, m2, m3, m4, m5)
s.calculate()
```

```python
class D:
    def showD(self):
        print("showD function")

class A(D):
    def showA(self):
        print("showA function")

class B(D):
    def showB(self):
        print("showB function")

class C(D):
    def showC(self):
        print("showC function")

obj1 = A()
obj2 = B()
obj3 = C()

obj1.showD()
obj1.showA()

obj2.showD()
obj2.showB()

obj3.showD()
obj3.showC()
```
